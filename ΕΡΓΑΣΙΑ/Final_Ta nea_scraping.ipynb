```python
import requests
from bs4 import BeautifulSoup
import pandas as pd # Import pandas
import time

full_articles_data = [] # List to store dictionaries of article data

start_page = 1
end_page = 7
base_tag_url = 'https://www.tanea.gr/tag/%CF%85%CF%80%CE%BF%CE%BA%CE%BB%CE%BF%CF%80%CE%AD%CF%82/'
main_url = 'https://www.tanea.gr'

for page_num in range(start_page, end_page + 1):
    print(f"--- Processing Page {page_num} ---")

    if page_num == 1:
        page_url = base_tag_url
    else:
        page_url = f"{base_tag_url}page/{page_num}/"

    try:
        response_page = requests.get(page_url)
        response_page.raise_for_status() # Raise an HTTPError for bad responses (4xx or 5xx)
        doc = BeautifulSoup(response_page.text, 'html.parser') # Update doc for each page
    except requests.exceptions.RequestException as e:
        print(f"Could not fetch page {page_url}: {e}")
        continue # Skip to the next page if page fetch fails

    article_link_elements = doc.find_all('a', class_='is-block is-relative row-box')

    if not article_link_elements:
        print(f"No article links found on page {page_num}. Skipping.")
        continue

    for article_link_element in article_link_elements:
        story_dict = {} # Dictionary to hold data for the current article
        story_dict['site'] = 'tanea.gr'

        # Get the URL directly from the found link element
        article_href = article_link_element['href']
        if article_href.startswith('http'):
            article_url = article_href
        else:
            article_url = main_url + article_href # Ensure full URL
        story_dict['url'] = article_url

        # The parent div of the link element should be the article teaser container
        article_element = article_link_element.find_parent('div', class_=lambda x: x and 'column is-full' in x)
        if not article_element:
            article_element = article_link_element # Fallback

        headline = article_element.find('h3')
        story_dict['headline'] = headline.text.strip() if headline else ''

        category_element = article_element.find('span', {'class': 'is-bold is-size-4-desktop is-size-5-touch'})
        story_dict['category'] = category_element.text.strip() if category_element else ''

        author_element = article_element.find('span', {'class': 'is-semibold mr-4 uppercase is-size-5 is-neablue mb-1'})
        story_dict['author'] = author_element.text.strip() if author_element else ''

        date_element = article_element.find('span', {'class': 'is-size-5 is-darkblue'})
        story_dict['date'] = date_element.text.strip() if date_element else ''

        # Fetch the full article content
        try:
            response_article = requests.get(article_url)
            response_article.raise_for_status()
            article_doc = BeautifulSoup(response_article.text, 'html.parser')

            main_content_div = article_doc.find('div', {'class': 'post-body main-content pos-rel article-wrapper'})
            if main_content_div:
                paragraphs = main_content_div.find_all('p')
                text = '\n'.join([p.get_text(strip=True) for p in paragraphs if p.get_text(strip=True)])
            else:
                text = "Main article content div not found."
            story_dict['full_text'] = text

        except requests.exceptions.RequestException as e:
            story_dict['full_text'] = f"Error fetching full text: {e}"
        except Exception as e:
            story_dict['full_text'] = f"Error processing full text: {e}"

        full_articles_data.append(story_dict) # Add the story dictionary to the list

    time.sleep(1) # Be respectful to the server

# After the loop, create a Pandas DataFrame
df = pd.DataFrame(full_articles_data)
print("\nFinished scraping. DataFrame head:")
print(df.head())
```
