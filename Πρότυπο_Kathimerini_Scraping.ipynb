```python
import requests
from bs4 import BeautifulSoup
```

```python
response = requests.get ('https://www.kathimerini.gr/politics/')
doc = BeautifulSoup(response.text, 'html.parser')
```

```python
titles = doc.find_all('p',{'class':'title is-5 medium_title'})

for t in titles:
  print(t.text)
```

```python
stories = doc.find_all('div', {'class': 'card-content column'})
for s in stories:
  title = s.find('p', {'class': 'title is-5 medium_title'})
  if title:
    print(title.text)

  # Find the anchor tag within the current story `s`
  link_tag = s.find('a', href=True)

  if link_tag:
    article_href = link_tag['href']
    # Check if the href is already a full URL or needs to be prepended
    if article_href.startswith('http'):
        article_url = article_href
    else:
        article_url = "https://www.kathimerini.gr" + article_href

    print(article_url)
    article_response = requests.get(article_url)
    article_doc = BeautifulSoup(article_response.text, 'html.parser')

    article_content = article_doc.find_all('p')
    text = " "
    for content in article_content:
      print(content.text)
      text = text + content.text
    div_article = article_doc.find('div', { 'id': 'articleBody' })
    print(div_article)
  else:
    print("Link not found for this story.")
```

```python
stories_list = []

for i in range(1,4):
  url = 'https://www.kathimerini.gr/politics/page/' + str(i) + '/'
  print(f"Fetching URL: {url}")
  response = requests.get(url)
  doc = BeautifulSoup(response.text, 'html.parser')

  # Find all article containers
  stories = doc.find_all('div', {'class': 'card columns is-flex-direction-row cat-list-rtl m-0'})


  for story in stories:
    story_data = {}
    print("--- Found a story ---")

    # Extract Headline
    headline_tag = story.find('p', class_='title is-5 medium_title')
    if headline_tag:
        story_data['headline'] = headline_tag.text.strip()
        print(f"Headline: {story_data['headline']}")


    # Extract Link
    link_tag = story.find('div', class_='card-content').find('a', href=True)
    if link_tag:
        story_data['link'] = link_tag['href']
        if not story_data['link'].startswith('http'):
            story_data['link'] = "https://www.kathimerini.gr" + story_data['link']
        print(f"Link: {story_data['link']}")
    else:
        story_data['link'] = None
        print("Link not found.")

    # Extract Summary
    summary_div = story.find('div', class_='content is-5 medium_lead is-hidden-touch')
    if summary_div and summary_div.find('p'):
        story_data['summary'] = summary_div.find('p').text.strip()
        print(f"Summary: {story_data['summary']}")
    else:
        story_data['summary'] = None
        print("Summary not found.")

    #Content:
    if story_data['link']:
        article_url = story_data['link'] # Corrected: use the extracted link for the article
        print(f"Fetching article content from: {article_url}")
        try:
            article_response = requests.get(article_url)
            article_doc = BeautifulSoup(article_response.text, 'html.parser')

            article_content_div = article_doc.find('div', {'id': 'articleBody'})
            text = " "
            if article_content_div:
                paragraphs = article_content_div.find_all('p')
                for p in paragraphs:
                    text += p.text.strip() + "\n"
            else:
                # Fallback if 'articleBody' div is not found
                paragraphs = article_doc.find_all('p')
                for p in paragraphs:
                    text += p.text.strip() + "\n"
            story_data['full_content'] = text.strip()
            # print(f"Full Content: {story_data['full_content'][:100]}...") # Print first 100 chars
        except requests.exceptions.RequestException as e:
            print(f"Error fetching article content for {article_url}: {e}")
            story_data['full_content'] = 'Error fetching content.'
    else:
        story_data['full_content'] = 'No link available to fetch content.'

    stories_list.append(story_data)

print(f"Total stories collected: {len(stories_list)}")
```

```python
import pandas as pd

if stories_list:
  df_kathimerini = pd.DataFrame(stories_list)
  df_kathimerini.index = range(1, len(df_kathimerini) + 1)
  display(df_kathimerini)
```
